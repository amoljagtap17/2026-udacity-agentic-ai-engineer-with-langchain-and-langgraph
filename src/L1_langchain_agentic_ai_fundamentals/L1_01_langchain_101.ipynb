# LangChain 101

**Load Environment Variables**

```python
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
```

```python
load_dotenv()
```

```python
model = init_chat_model(
    "gpt-5.6-sol",
    # Kwargs passed to the model:
    temperature=0.0,
    timeout=30,
    max_tokens=1000,
    max_retries=2,  # Default; increase for unreliable networks
)
```

```python
response = model.invoke("What is the latest react version you are trained on?")
```

```python
response
```

```python
print(response)
```

```python
response.content
```

```python
response.usage_metadata
```

```python
from langchain_core.load import dumpd, load
```

```python
# Serialize to a plain dict
serialized = dumpd(response)

serialized
```

```python
# Deserialize back to a message object
restored = load(serialized)

restored
```

**Using Chat Models**

```python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(
    model="gpt-5.6-sol",
    # stream_usage=True,
    # temperature=None,
    # max_tokens=None,
    # timeout=None,
    # reasoning_effort="low",
    # max_retries=2,
    # api_key="...",  # If you prefer to pass api key in directly
    # base_url="...",
    # organization="...",
    # other params...
)
```

```python
messages = [
    (
        "system",
        "You are a helpful assistant.",
    ),
    ("human", "Hello There!"),
    ("ai", "Hello! I'm Agent AJ. How can I assist you today?"),
]
ai_msg = llm.invoke(messages)
ai_msg
```

```python
ai_msg.pretty_print()
```

```python
messages.append(ai_msg)
```

```python
messages
```

```python
from langchain.messages import HumanMessage, SystemMessage, AIMessage
```

```python
messages = [
    SystemMessage(content="You are a helpful assistant."),
    HumanMessage(content="Hello There!"),
    AIMessage(content="Hello! I'm Agent AJ. How can I assist you today?"),
]
ai_msg = llm.invoke(messages)
ai_msg
```

```python
ai_msg.content
```

```python
ai_msg.content[-1]
```

```python
ai_msg.text
```

```python
messages.append(AIMessage(content=ai_msg.text))
```

```python
for message in messages:
    message.pretty_print()
```
