# LangChain Multistep Workflow

**Challenge**

Create an AI Business Advisor that:

1. Accepts an industry as input.
2. Generates a business idea.
3. Analyzes the strengths and weaknesses.
4. Formats the results as a final report.

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import (
    RunnablePassthrough,
    RunnableParallel,
    RunnableLambda,
)
from pydantic import BaseModel, Field
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
model = "gpt-5.6-sol"
```

```python
llm = ChatOpenAI(
    model=model,
    temperature=0.0,
    max_tokens=500,
    timeout=500,
    reasoning_effort="low",
    max_retries=2,
)
```

```python
logs = []
```

```python
parser = StrOutputParser()
```

```python
parse_and_log_output_chain = RunnableParallel(
    output=parser, log=RunnableLambda(lambda x: logs.append(x))
)
```

```python
idea_prompt = PromptTemplate(
    template=(
        "You are creative business consultant."
        "Generate one innovative business idea for a startup based on the following industry:"
        "{industry}."
        "Provide a brief description of the idea and explain why it is innovative."
    )
)
```

```python
idea_chain = idea_prompt | llm | parse_and_log_output_chain
```

```python
idea_result = idea_chain.invoke("Fintech")
```

```python
idea_result["output"]
```

```python
for message in logs:
    print(message.content)
```

```python
analysis_prompt = PromptTemplate(
    template=(
        "Analyze the following business idea: "
        "Idea: {idea} "
        "Identify 3 key strengths and 3 potential weaknesses of the idea."
    )
)
```

```python
analysis_chain = analysis_prompt | llm | parse_and_log_output_chain
```

```python
analysis_result = analysis_chain.invoke(idea_result["output"])
```

```python
analysis_result["output"]
```

```python
logs
```

```python
for message in logs:
    print(message.content)
```

```python
for message in logs:
    message.pretty_print()
```

```python
report_prompt = PromptTemplate(
    template=(
        "Here is a business analysis: "
        "Strengths & Weaknesses: {output} "
        "Generate a structured business report."
    )
)
```

```python
class AnalysisReport(BaseModel):
    """Strengths and Weaknesses about a business idea"""

    strengths: list = Field(default=[], description="Idea's strength list")
    weaknesses: list = Field(default=[], description="Idea's weaknesse list")
```

```python
report_chain = report_prompt | llm.with_structured_output(
    schema=AnalysisReport, method="function_calling"
)
```

```python
report_result = report_chain.invoke(analysis_result["output"])
```

```python
report_result
```

```python
report_result.strengths
```

```python
report_result.weaknesses
```

```python
e2e_chain = (
    RunnablePassthrough()
    | idea_chain
    | RunnableParallel(idea=RunnablePassthrough())
    | analysis_chain
    | report_chain
)
```

```python
e2e_chain.get_graph().print_ascii()
```

```python
e2e_result = e2e_chain.invoke("Finance")
```
