**Messages**

```python
from dotenv import load_dotenv
from langchain.messages import AIMessage, HumanMessage, SystemMessage
from langchain_openai import ChatOpenAI
```

```python
load_dotenv()
```

```python
model = "gpt-5.6-sol"
```

```python
llm = ChatOpenAI(
    model=model,
    temperature=0.0,
    max_tokens=500,
    timeout=500,
    reasoning_effort="low",
    max_retries=2,
)
```

```python
llm.invoke("Hello there")
```

```python
messages = [
    SystemMessage("You are a geography tutor"),
    HumanMessage("What's the capital of Australia?"),
]
```

```python
llm.invoke(messages)
```

```python
messages = [
    SystemMessage("You are a geography tutor"),
    HumanMessage("What's the capital of Australia?"),
    AIMessage("The capital of Australia is Canberra"),
    HumanMessage("What's the capital of USA?"),
]
```

```python
response = llm.invoke(messages)
```

**Few Shot Prompting**

```python
example_prompt = PromptTemplate(
    template="Question: {input}\nThought: {thought}\nResponse: {output}"
)
```

```python
examples = [
    {
        "input": "What is Python?",
        "thought": "Python is a high-level programming language.",
        "output": "Python is a high-level programming language.",
    },
    {
        "input": "What is React?",
        "thought": "React is a JavaScript library for building user interfaces.",
        "output": "React is a JavaScript library for building user interfaces.",
    },
    {
        "input": "What is JavaScript?",
        "thought": "JavaScript is a programming language commonly used for web development.",
        "output": "JavaScript is a programming language commonly used for web development.",
    },
]
```

```python
print(example_prompt.invoke(examples[0]).to_string())
```

```python
from langchain_core.prompts import FewShotPromptTemplate
```

```python
prompt_template = FewShotPromptTemplate(
    example_prompt=example_prompt,
    examples=examples,
    suffix="Question: {input}",
    input_variables=["input"],
    example_separator="\n\n",
)
```

```python
response = llm.invoke(prompt_template.invoke({"input": "What is JavaScript?"}))
```

```python
response
```

```python
response.content
```

**Chat Prompt Templates**

```python
from langchain_core.prompts import ChatPromptTemplate
```

```python
template = ChatPromptTemplate(
    [
        ("system", "You are a helpful AI bot. Your name is {name}."),
        ("human", "Hello, how are you doing?"),
        ("ai", "I'm doing well, thanks!"),
        ("human", "{user_input}"),
    ]
)
```

```python
prompt_value = template.invoke(
    {
        "name": "Bob",
        "user_input": "What is your name?",
    }
)
```

```python
llm.invoke(prompt_value)
```

```python
template = ChatPromptTemplate(
    [
        ("system", "You are a helpful AI bot."),
        # Means the template will receive an optional list of messages under
        # the "conversation" key
        ("placeholder", "{conversation}"),
        # Equivalently:
        # MessagesPlaceholder(variable_name="conversation", optional=True)
    ]
)
```

```python
prompt_value = template.invoke(
    {
        "conversation": [
            ("human", "Hi!"),
            ("ai", "How can I assist you today?"),
            ("human", "Can you make me an ice cream sundae?"),
            ("ai", "No."),
        ]
    }
)
```

```python
prompt_value
```

```python
prompt_value.messages
```

```python
for message in prompt_value.messages:
    message.pretty_print()
```
