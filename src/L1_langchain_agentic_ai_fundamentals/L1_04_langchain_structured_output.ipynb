# Langchain Structured Output

- Pydantic models: BaseModel subclasses with field validation. Returns validated Pydantic instance.
- Dataclasses: Python dataclasses with type annotations. Returns dict.
- TypedDict: Typed dictionary classes. Returns dict.

```python
from dotenv import load_dotenv
from langchain.messages import HumanMessage, SystemMessage
from langchain_openai import ChatOpenAI
from pydantic import BaseModel, Field
```

```python
load_dotenv()
```

```python
model = "gpt-5.6-sol"
```

```python
llm = ChatOpenAI(
    model=model,
    temperature=0.0,
    max_tokens=500,
    timeout=500,
    reasoning_effort="low",
    max_retries=2,
)
```

**Pydantic**

```python
class ContactInfo(BaseModel):
    """Contact information for a person."""

    name: str = Field(description="The name of the person")
    email: str = Field(description="The email address of the person")
    phone: str = Field(description="The phone number of the person")
```

```python
system_message = SystemMessage(
    content="You are a helpful assistant that extracts contact information from text."
)
```

```python
human_message = HumanMessage(
    content="Extract the contact information from the following text: 'John Doe, email: john.doe@example.com, phone: 123-456-7890'"
)
```

```python
llm_with_structure = llm.with_structured_output(ContactInfo)
```

```python
response = llm_with_structure.invoke([system_message, human_message])
```

```python
response
```

**Dataclasses**

```python
from dataclasses import dataclass
```

```python
@dataclass
class ContactInfo:
    """Contact information for a person."""

    name: str  # The name of the person
    email: str  # The email address of the person
    phone: str  # The phone number of the person
```

```python
llm_with_structure = llm.with_structured_output(ContactInfo)
```

```python
response = llm_with_structure.invoke([system_message, human_message])
```

```python
response
```

**TypedDict**

```python
from typing_extensions import TypedDict
```

```python
class ContactInfo(TypedDict):
    """Contact information for a person."""

    name: str  # The name of the person
    email: str  # The email address of the person
    phone: str  # The phone number of the person
```

```python
llm_with_structure = llm.with_structured_output(ContactInfo)
```

```python
response = llm_with_structure.invoke([system_message, human_message])
```
