# LangChain Chatbot Application

```python

from dotenv import load_dotenv
from langchain_core.messages import AIMessage, HumanMessage, SystemMessage
from langchain_core.prompts import (
    ChatPromptTemplate,
    FewShotChatMessagePromptTemplate,
)
from langchain_openai import ChatOpenAI
```

```python
load_dotenv()
```

```python
model = "gpt-5.6-sol"
```

```python
class ChatBot:
    def __init__(
        self,
        name: str,
        instructions: str,
        examples: list[dict],
        model: str = model,
        temperature: float = 0.0,
    ):

        self.name = name

        self.llm = ChatOpenAI(
            model=model,
            temperature=temperature,
            max_tokens=500,
            timeout=50,
            reasoning_effort="low",
            max_retries=2,
        )

        messages = [
            SystemMessage(content=instructions),
            HumanMessage(content="{input}"),
            AIMessage(content="{output}"),
        ]

        example_prompt = ChatPromptTemplate.from_messages(messages)

        prompt_template = FewShotChatMessagePromptTemplate(
            example_prompt=example_prompt,
            examples=examples,
        )

        # Memory
        self.messages = prompt_template.invoke({}).to_messages()

    def invoke(self, user_message: str) -> AIMessage:
        self.messages.append(HumanMessage(user_message))
        ai_message = self.llm.invoke(self.messages)
        self.messages.append(ai_message)
        return ai_message
```

```python
instructions = (
    "You are a helpful assistant with extensive knowledge of the LangChain framework."
    "You will use tone and style that is professional, informative, and approachable."
    "You will provide clear and concise explanations, examples, and guidance to users seeking assistance with LangChain."
    "You will answer questions about LangChain, including its features, installation, usage, and best practices."
    "For any questions outside the scope of LangChain, you will politely inform the user that you are unable to provide assistance."
)
```

```python
examples = [
    {
        "input": "What is LangChain?",
        "output": "LangChain is a framework for developing applications powered by language models.",
    },
    {
        "input": "How do I install LangChain?",
        "output": "You can install LangChain using pip: pip install langchain",
    },
]
```

```python
bot = ChatBot(name="LangChainBot", instructions=instructions, examples=examples)
```

```python
response = bot.invoke("What is React?")
```
